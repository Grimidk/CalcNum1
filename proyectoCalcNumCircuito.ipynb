```python
import numpy as np
import matplotlib.pyplot as plt

def rungeKutta(f, t0, tf, y0, h):

    t = t0
    y = y0
    t_values = [t0]
    y_values = [y0]
    while t < tf:
        k1 = h * f(t, y)
        k2 = h * f(t + h/2, y + k1/2)
        k3 = h * f(t + h/2, y + k2/2)
        k4 = h * f(t + h, y + k3)
        y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
        t = t + h
        t_values.append(t)
        y_values.append(y)
    return np.array(t_values), np.array(y_values)

def auxiliar(i, t):
    return (w * x * np.cos(w * t) - c*10**(-6) * i) / r

w = 377
c = 10.857
r = 29.08
x = 55
t0 = 0
tf = 1 
y0 = 0
h = 0.0001 
t_values, i_values = rungeKutta(auxiliar, t0, tf, y0, h)

plt.figure(figsize=(10, 6))
plt.plot(t_values, i_values, label='Solución de i(t)')
plt.title('Solución de la Ecuación Diferencial con Runge-Kutta')
plt.xlabel('tiempo')
plt.ylabel('i(t)')
plt.legend()
plt.grid(True)
plt.show()
```
